__1. Problem1__: A fair coin is tossed 100 times.Calculate: 

* Number of heads  

* Number of tails 

* Experimental probability of Head  

* Experimental probability of Tail 

```python
import random

heads = 0
tails = 0
n = 100

for _ in range(n):
    result = random.choice(["Heads", "Tails"])

    if result == "Heads":
        heads += 1
    else:
        tails += 1

print("Number of heads : ", heads)
print("number of tails : ", tails)
print("\nExperimental probability of heads : ",heads/n)
print("Experimental probabiliy of tails : ", tails/n)
```

__3. Rolling a die:__ A fair die is rolled 6,000 times. 

* Calculate the experimental probability of each outcome: S={1,2,3,4,5,6} 

```python
import random

n = 6000
total = 0
times = 0
die = [0, 0, 0, 0, 0, 0]

for _ in range(n):
    roll = random.randint(1, 6)

    die[roll - 1] += 1

for i in range(6):
    die_number = i + 1 
    count = die[i]
    probability = round(die[i]/n, 5)

    print(f"Die : {die_number} rolled : {count} times | Experimental probaility : {probability}")
    total += probability
    times += count

print("\nVerification : ")
print(f"Total probaility :  {total:.2f}")
print(f"Times : {times}")
```

__4. Addition rule for disjoint events__ 

* A die is rolled 10,000 times. 

* Define: A=getting 1, B= getting 6, check the events are joint or disjoint  

* Verify: P(A∪B)=P(A)+P(B)  

```python
import random

n = 10000

count_A = 0
count_B = 0
count_either = 0
count_both = 0


for _ in range(n):
    roll = random.randint(1, 6)

    if roll == 1:
        count_A += 1
        count_either += 1
    elif roll == 6:
        count_B += 1
        count_either += 1

    if roll == 1 and roll == 6:
        count_both += 1


print("Probability of event A : ", count_A/n)
print("Probability of event B : ", count_B/n)
print("Probability of both happening together : ", count_both/n)

if count_both == 0:
    print("\nConclusion : The events are disjoint")
else : 
    print("\nConclusion : The events are joint")


A_union_B = count_either/n

prob_A_plus_prob_B = (count_A + count_B)/n

print("\nVerifying: P(A∪B)=P(A)+P(B)...  ")
if A_union_B == prob_A_plus_prob_B:
    print("P(A U B) is equal to P(A) + P(B)")
else:
    print("P(A U B) is equal to P(A) + P(B)")
```

__5. Addition rule for overlapping events__

* For one die roll, define: A=getting an even number A={2,4,6} and B=getting a number greater than 3 B={4,5,6} 

* Verify: P(A∪B)=P(A)+P(B)−P(A∩B) 

```python
import random

n = 1
A = []
B = []
A_and_B = []
A_or_B = []
count_neither = []

roll = random.randint(1, 6)
print(f"The die landed on : {roll}")

set_A = {2, 4, 6}
set_B = {4, 5, 6}

if roll in set_A:
    A.append(roll)

if roll in set_B:
    B.append(roll)

if roll in set_A and roll in set_B:
    A_and_B.append(roll)

if roll in set_A or roll in set_B:
    A_or_B.append(roll)
else:
    count_neither.append(roll)


prob_A = len(A)/ n
prob_B = len(B)/ n
prob_intersection = len(A_and_B)/ n
prob_union = len(A_or_B)/ n
prob_neither= len(count_neither)/ n

print("Probaility of A: ", prob_A)
print("Probaility of B: ", prob_A)
print("Probaility of (A ∩ B): ", prob_intersection)
print("Probaility of (A U B): ", prob_union)
print("Probaility of neither: ", prob_neither)



print("List A (getting an even number) : ", A)
print("List B ( getting a number greater than 3) :", B)
print("List Both ( A ∩ B ) : ", A_and_B)
print("List Either (A or B) : ", A_or_B)
print("List neither : ", count_neither)


print("\nVerifying: P(A U B)=P(A) + P(B) - P(A ∩ B)  ")
if prob_union == prob_A + prob_B - prob_intersection:
    print("\nP(A U B) is equal  to P(A) + P(B) - P(A ∩ B)")
else:
    print("\nP(A U B) is equal to P(A) + P(B) - P(A ∩ B)")
```

__6. Multiplication rule for independent events__

* A coin is tossed and a die is rolled 10,000 times. 

* Define: 

    - obtaining Head  

    - obtaining an even number 

    - Verify: P(A∩B)=P(A)P(B) 

```python
import random

n = 10000
A = []
B = []
A_and_B = []

for _ in range(n):
    coin = random.choice(['H', 'T'])
    roll = random.randint(1, 6)

    if coin == 'H':
        A.append(coin)

    if roll in {2, 4, 6}:
        B.append(roll)

    if coin == 'H' and roll in [2, 4, 6]:
        A_and_B.append((coin, roll))


prob_A = len(A)/n
prob_B = len(B)/n
prob_intersection = len(A_and_B)/n

print("Probabilty of A (Head) : ", prob_A)
print("Probability of B (Even) : ", prob_B)
print("Proability of (A ∩ B) : ", prob_intersection)

print("\nVerifying Independence : P(A ∩ B) = P(A) * P(B)")

rhs = prob_A * prob_B

print("LHS : ", round(prob_intersection, 1))
print("RHS : ", round(rhs, 1))

if round(prob_intersection, 1) == round(rhs, 1):
    print("\nP(A ∩ B) is equal to P(A) * P(B)")
else:
    print("\nP(A ∩ B) is not equal to P(A) * P(B)")
```
